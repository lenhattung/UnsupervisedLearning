# Demo: K-Means thất bại với moon dataset

```python
from sklearn.datasets import make_moons
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Tạo moon dataset - 2 cụm hình bán nguyệt
X_moons, y_moons = make_moons(n_samples=300, noise=0.05, random_state=42)

# Thử K-Means
kmeans_moons = KMeans(n_clusters=2, random_state=42)
y_pred_kmeans = kmeans_moons.fit_predict(X_moons)

```

```python
# Vẽ so sánh
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Ground truth
axes[0].scatter(X_moons[:, 0], X_moons[:, 1], c=y_moons, cmap='viridis')
axes[0].set_title('Ground Truth - 2 cụm hình bán nguyệt')

# K-Means result
axes[1].scatter(X_moons[:, 0], X_moons[:, 1], c=y_pred_kmeans, cmap='viridis')
axes[1].scatter(kmeans_moons.cluster_centers_[:, 0],
                kmeans_moons.cluster_centers_[:, 1],
                s=300, c='red', marker='X')
axes[1].set_title('K-Means - Thất bại!')

plt.tight_layout()
plt.show()

# K-Means cắt dọc thay vì theo hình bán nguyệt!
```
