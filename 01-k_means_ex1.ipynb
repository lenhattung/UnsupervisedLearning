```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
```

```python
# Tạo dữ liệu mẫu
X, y_true = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=0.60,
    random_state=42
)
```

```python
# Khởi tạo và huấn luyện
kmeans = KMeans(
 n_clusters=4,
 init='k-means++',
 n_init=10,
 random_state=42
)
kmeans.fit(X)
```

```python
# Dự đoán cụm
labels = kmeans.predict(X)
centroids = kmeans.cluster_centers_
```

```python
# Vẽ kết quả
plt.figure(figsize=(10, 6))

# Vẽ các điểm dữ liệu
plt.scatter(
    X[:, 0], X[:, 1],
    c=labels,
    cmap='viridis',
    marker='o',
    alpha=0.6
)

# Vẽ centroids
plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    c='red',
    marker='X',
    s=200,
    edgecolors='black'
)

plt.title('K-Means Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```

```python
# Tính Inertia (WCSS)
inertia = kmeans.inertia_
print(f"Inertia: {inertia}")

# Silhouette Score
from sklearn.metrics import (
    silhouette_score
)
score = silhouette_score(X, labels)
print(f"Silhouette: {score}")
```
